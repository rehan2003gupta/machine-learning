## Pandas

Pandas is a fast, powerful, flexible and easy to use open source data analysis and manipulation tool, built on top of the Python programming language.

### Pandas Series

A Pandas Series is like a column in a table. It is a 1-D array holding data of any type.

Importing pandas 

```python
import pandas as pd
import numpy as np
```

### Series from lists 

```python
#string
country = ['usa', 'canada', 'mexico','india']
pd.Series(country)
```

```python
#integer
ages = [25, 30, 35, 40]
pd.Series(ages)
```

```python
#custom index
runs= [11,16,6,149,97,45]
cricketers = ['dhoni','kohli','sachin','raina','rohit','yuvi']
run_scored=pd.Series(runs, index=cricketers)
run_scored
```

```python
pd.Series(runs, index=cricketers, name='Runs Scored')
```

### Series from Dictionaries

```python
marks ={
    'english': 90,
    'maths': 95,
    'hindi': 85,
    'science': 99,
}
marks_scored = pd.Series(marks,name='Marks Scored')
marks_scored
```
