### Advanced Indexing

```python
#Normal indexing
a= np.arange(16).reshape(4,4)
```

```python
a
```

```python
a[1,0:2]
```

Fancy Indexing provide ability to access whatever you want in any order you want. It is a way to index an array using another array of indices.

```python
a[:, [0,2,3]]
```

Boolean indexing is a way to index an array using another array of boolean values. The boolean array must have the same shape as the array being indexed. The result of boolean indexing is a new array that contains only the elements of the original array that correspond to True values in the boolean array.

```python
#Boolean indexing
a = np.random.randint(1,100,24).reshape(6,4)
a
```

```python
#greater than 50
a[a>50]
```

```python
a>50
```

```python
(a>50) & (a%2==0)
```

```python
a[(a>50) & (a%2==0)]
```

### Broadcasting
Broadcasting is a powerful mechanism that allows numpy to work with arrays of different shapes when performing arithmetic operations. It automatically expands the smaller array to match the shape of the larger array, enabling element-wise operations without the need for explicit replication of data.

```python
a = np.arange(16).reshape(4,4)
b = np.arange(16,32).reshape(4,4)
c=a+b
print(c)
```

```python
a = np.arange(16).reshape(4,4)
b = np.arange(16,20).reshape(1,4)
c = a+b
print(a)
print(" ")
print(b)
print(" ")
print(c)
```

### Broadcasting Rules
1. If the arrays do not have the same number of dimensions, prepend the shape of the smaller array with ones until both shapes have the same length.
2. The two arrays are said to be compatible in a dimension if they have the same size in that dimension, or if one of the arrays has size 1 in that dimension.


```python
a = np.arange(12).reshape(3,4)
b = np.arange(3)
print(a)

print(b)
#print(a+b)
```

```python
b= np.arange(3).reshape(3,1)
print(a+b)
```

```python
#2nd example
a= np.arange(3).reshape(1,3)
print(a)
b= np.arange(3).reshape(3,1)
print(b)
print(a+b)
```

```python
a = np.array([1])
print(a)
b = np.arange(4).reshape(2,2)
print(b)
print(a+b)
```

```python
a = np.arange(16).reshape(4,4)
b = np.arange(4).reshape(2,2)
print(a)
print(b)    
#print(a+b)
# is breaking the rules
```

Mean Squared Error

```python
actual = np.random.randint(1,50,25)
predicted = np.random.randint(1,50,25)
actual
```

```python
predicted 
```

```python
actual - predicted
```

```python
def mse(actual, predicted):
    return np.mean((actual - predicted)**2)

mse(actual, predicted)
```
