5. Names parameter

 While importing data using pandas, there is an attribute called names using which we can specify the column names of the data. If the data has no header, we can use this attribute to specify the column names.

6. Index_col parameter

 The index_col parameter is used to specify the column that should be used as the index of the DataFrame. By default, pandas will create a new integer index for the DataFrame, but you can use the index_col parameter to specify a column to use as the index instead.

7. Header parameter

 The header parameter is used to specify the row that should be used as the header of the DataFrame. By default, pandas will use the first row of the data as the header, but you can use the header parameter to specify a different row to use as the header instead.

8. use_cols parameter

 The usecols parameter is used to specify a subset of columns to read from the data. By default, pandas will read all columns from the data, but you can use the usecols parameter to specify a list of column names or indices to read instead.

9. Squeeze parameter

 The squeeze parameter is used to convert a DataFrame with a single column into a Series. By default, pandas will return a DataFrame even if it has only one column, but you can use the squeeze parameter to return a Series instead.

10. Skiprows parameter

 The skiprows parameter is used to skip a specified number of rows from the beginning of the data. By default, pandas will read all rows from the data, but you can use the skiprows parameter to skip a specified number of rows instead.

11. nrows parameter

 The nrows parameter is used to specify the number of rows to read from the data. By default, pandas will read all rows from the data, but you can use the nrows parameter to read a specified number of rows instead.

```python
pd.read_csv(r'C:\Users\RADHE\machine-learning\Rehan\DATASET\bollywood.csv', nrows=100, sep=',')
```

12. Encoding parameter

 The encoding parameter is used to specify the encoding of the data. By default, pandas will use the system's default encoding, but you can use the encoding parameter to specify a different encoding if needed.

```python
pd.read_csv(r'C:\Users\RADHE\machine-learning\Rehan\DATASET\bollywood.csv', encoding='latin-1')
```

13. Skip bad lines parameter

 The skip_bad_lines parameter is used to skip lines that are malformed or contain errors. By default, pandas will raise an error if it encounters a bad line, but you can use the skip_bad_lines parameter to skip those lines instead.

 Example: pd.read_csv('data.csv', skip_bad_lines=True)

 Suppose you have 8 columns in all of the rows, but in some rows, there are only 7 columns. In that case, pandas will raise an error while reading the data. To avoid this error, you can use the skip_bad_lines parameter to skip those rows with missing columns.

14. dtypes parameter

 The dtypes parameter is used to specify the data types of the columns in the DataFrame. By default, pandas will infer the data types of the columns based on the data, but you can use the dtypes parameter to specify the data types explicitly.

 Example: pd.read_csv('data.csv', dtype={'column1': 'int64', 'column2': 'float64'})

15. Converters parameter

 The converters parameter is used to specify a dictionary of functions to apply to the columns of the DataFrame. By default, pandas will not apply any functions to the columns, but you can use the converters parameter to apply custom functions to the columns.

 Example: pd.read_csv('data.csv', converters={'column1': lambda x: x.upper(), 'column2': lambda x: float(x)})

```python
def rename(name):
    name = name.strip()

    if name == "Royal Challengers Bengaluru":
        return "RCB"

    return name

IPL = pd.read_csv(
    r'C:\Users\RADHE\machine-learning\Rehan\DATASET\IPL.csv',
    converters={'batting_team': rename}
)
IPL.tail(100)
```

16. na_values parameter

 The na_values parameter is used to specify a list of values that should be treated as missing or NaN in the DataFrame. By default, pandas will treat certain values (such as 'NaN', 'null', etc.) as missing, but you can use the na_values parameter to specify additional values to treat as missing.

 Example: pd.read_csv('data.csv', na_values=['NA', 'missing', 'n/a'])

17. low memory parameter

 The low_memory parameter is used to optimize memory usage when reading large datasets. By default, pandas will read the entire dataset into memory, which can be inefficient for large datasets. Setting low_memory=True allows pandas to read the data in smaller chunks, which can help reduce memory usage.

 Example: pd.read_csv('data.csv', low_memory=True)

18. Loading a huge dataset in chunks

 When working with large datasets, it's often necessary to load the data in smaller chunks rather than all at once. This can help manage memory usage and improve performance. The `chunksize` parameter in `pd.read_csv()` allows you to specify the number of rows to read at a time.

 Example: pd.read_csv('data.csv', chunksize=1000)
